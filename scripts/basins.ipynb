```python
salina_basins = na_basins[na_basins.HYBAS_ID.isin(basin_ids)]
```

```python
salina_basins.explore(figsize=(20, 10))
```

### Counties intersecting basin

```python
counties = gpd.read_file("../data/raw/admin/cb_2018_us_county_500k.zip")
```

```python
counties = counties.to_crs(salina_basins.crs)
```

```python
intersections = counties.sindex.query(salina_basins.geometry, predicate='intersects')
```

```python
intersections
```

```python
basin_counties = counties.iloc[intersections[1]]
```

```python
basin_counties.explore()
```

```python
m = basin_counties.explore()
salina_basins.explore(m=m, color='yellow')
```

```python
salina_basins = salina_basins.to_crs(salina_basins.estimate_utm_crs())
```

```python
salina_basins.area.sum() # 1.1 million square km
salina_basins.to_file('../data/processed/basins/salina_basins.gpkg')
```
